### Proximity Measure (Dissimilarity) for Mixed-Type Attributes using Gower's Distance

Gower's distance is a metric that can handle a mix of categorical (nominal, ordinal, binary) and numerical variables. It calculates a weighted average of partial dissimilarities for each variable type.

The formula you provided, $ D_{ij} = \frac{\sum_{f=1}^{p} \delta_{ijf} d_{ijf}}{\sum_{f=1}^{p} \delta_{ijf}} $, is the core of Gower's distance:
*   $D_{ij}$ is the total dissimilarity between objects $i$ and $j$.
*   $p$ is the number of comparable features (attributes).
*   $\delta_{ijf}$ is a weight (usually 1 if the feature $f$ is comparable between $i$ and $j$, and 0 if not, e.g., due to missing values). In our implementation, we'll implicitly set this to 1 for all features we process.
*   $d_{ijf}$ is the partial dissimilarity between objects $i$ and $j$ for feature $f$.

Here's how partial dissimilarities ($d_{ijf}$) are typically calculated for different attribute types:
*   **Numerical**: Absolute difference divided by the range of the attribute.
*   **Binary (Symmetric/Asymmetric)**: 0 if values are the same, 1 if different (often treated as a special case of nominal or numerical).
*   **Nominal**: 0 if values are the same, 1 if different.
*   **Ordinal**: Ranks are used, and then treated like numerical differences, scaled by the maximum possible difference.

```python
import pandas as pd

data = [
    [1, "Code A", "Excellent", 1, 45],
    [2, "Code B", "Fair",      0, 22],
    [3, "Code C", "Good",      1, 64],
    [4, "Code A", "Excellent", 0, 28]
]

df = pd.DataFrame(
    data,
    columns=["id", "A1", "A2", "A3", "A4"]
)
df
```

```python
ordinal = {
    "Poor": 1,
    "Fair": 2,
    "Good": 3,
    "Excellent": 4
}
```

```python
min_A4 = df["A4"].min()
max_A4 = df["A4"].max()
```

```python
def mixed_dissimilarity(row1, row2):

    d1 = 0 if row1["A1"] == row2["A1"] else 1


    r1 = (ordinal[row1["A2"]] - 1) / 3
    r2 = (ordinal[row2["A2"]] - 1) / 3

    d2 = abs(r1 - r2)

    d3 = 0 if row1["A3"] == row2["A3"] else 1

    n1 = (row1["A4"] - min_A4) / (max_A4 - min_A4)
    n2 = (row2["A4"] - min_A4) / (max_A4 - min_A4)

    d4 = abs(n1 - n2)


    D = (d1 + d2 + d3 + d4) / 4
    return D
```

```python
n = len(df)

dm = pd.DataFrame(
    0.0,
    index=df["id"],
    columns=df["id"]
)

for i in range(n):
    for j in range(n):

        D = mixed_dissimilarity(
            df.iloc[i],
            df.iloc[j]
        )

        dm.iloc[i, j] = D

print("\nMixed Dissimilarity Matrix:")
print(dm.round(2))
```
