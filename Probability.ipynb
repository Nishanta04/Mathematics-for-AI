```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")
df.head()
```

```python
pd.crosstab(df['Survived'], df['Pclass'])
```

##Joint Probabilities

```python
pd.crosstab(df['Survived'], df['Pclass'], normalize='all')
```

##Joint + Marginal Probabilities

```python
pd.crosstab(df['Survived'], df['Pclass'], normalize='all', margins=True)
```

##Conditional Probabilities acc. to columns P(A|B)

```python
pd.crosstab(df['Survived'], df['Pclass'], normalize='columns')
```

##Conditional Probabilities acc. to rows P(B|A)

```python
pd.crosstab(df['Survived'], df['Pclass'], normalize='index')
```
