## Question 1: Vector Operations


```python
import numpy as np
```

```python
# Define the vectors
a = np.array([2, 4, 6, 8])
b = np.array([1, 3, 5, 7])

print(f"Vector a: {a}")
print(f"Vector b: {b}")
```

### Vector Addition

```python
# Vector addition
vector_sum = a + b
print(f"Vector addition (a + b): {vector_sum}")
```

### Vector Subtraction

```python
# Vector subtraction
vector_difference = a - b
print(f"Vector subtraction (a - b): {vector_difference}")
```

### Dot Product

```python
# Dot product
dot_product = np.dot(a, b)
print(f"Dot product of a and b: {dot_product}")
```

### Euclidean Norm

```python
# Euclidean norm of each vector
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)
print(f"Euclidean norm of a: {norm_a:.4f}")
print(f"Euclidean norm of b: {norm_b:.4f}")
```

### Cosine Similarity

```python
# Cosine similarity
cosine_similarity = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
print(f"Cosine similarity between a and b: {cosine_similarity:.4f}")
```

## Question 2: Matrix Operations

```python
# Define the matrices
A = np.array([[2, 1, 3], [4, 5, 2], [1, 2, 4]])
B = np.array([[1, 2, 1], [3, 1, 2], [2, 4, 3]])

print("Matrix A:\n", A)
print("\nMatrix B:\n", B)
```

### Matrix Addition (A + B)

```python
# Calculate A + B
matrix_sum = A + B
print("A + B:\n", matrix_sum)
```

### Matrix Subtraction (A - B)

```python
# Calculate A - B
matrix_difference = A - B
print("A - B:\n", matrix_difference)
```

### Matrix Product (AB)

```python
# Calculate the matrix product AB
matrix_product = np.dot(A, B)
print("Matrix Product (AB):\n", matrix_product)
```

### Element-wise Product of A and B

```python
# Calculate the element-wise product of A and B
element_wise_product = A * B
print("Element-wise Product of A and B:\n", element_wise_product)
```

### Transpose of A

```python
# Calculate the transpose of A
transpose_A = A.T
print("Transpose of A:\n", transpose_A)
```

### Trace of A

```python
# Calculate the trace of A
trace_A = np.trace(A)
print("Trace of A:", trace_A)
```

## Question 3: Determinant and Inverse

```python
# Define the matrix A for Question 3
A_q3 = np.array([[4, 2, 1], [2, 5, 3], [1, 3, 6]])
print("Matrix A for Question 3:\n", A_q3)
```

### Determinant of A

```python
# Calculate the determinant of A
determinant_A = np.linalg.det(A_q3)
print(f"Determinant of A: {determinant_A:.4f}")
```

### Singular or Non-singular

```python
# Determine whether the matrix is singular or non-singular
# A matrix is singular if its determinant is zero.
if np.isclose(determinant_A, 0):
    print("The matrix A is singular.")
else:
    print("The matrix A is non-singular.")
```

### Inverse of A

```python
# Calculate the inverse of A
inverse_A = np.linalg.inv(A_q3)
print("Inverse of A:\n", inverse_A)
```

### Verify the Inverse (AA⁻¹)

```python
# Verify the inverse by calculating AA⁻¹
verification = np.dot(A_q3, inverse_A)
print("A * A⁻¹:\n", verification)
# Ideally, this should be close to the identity matrix.
```

## Question 4: Solving a System of Linear Equations

### Representing the System in AX = b

```python
# Represent the system of linear equations in the form AX = b
# A is the coefficient matrix
A_system = np.array([[2, 1, 1], [1, 3, 2], [1, 2, 4]])

# b is the constants vector
b_system = np.array([10, 13, 18])

print("Coefficient Matrix A:\n", A_system)
print("\nConstants Vector b:\n", b_system)
```

### Solve the System using `numpy.linalg.solve()`

```python
# Solve the system for X (x, y, z)
solution_X = np.linalg.solve(A_system, b_system)

print(f"Solution (x, y, z): {solution_X}")
print(f"x = {solution_X[0]:.4f}")
print(f"y = {solution_X[1]:.4f}")
print(f"z = {solution_X[2]:.4f}")
```

### Verify the Solution

```python
# Verify the solution by substituting the obtained values back into the original equations
x, y, z = solution_X

eq1_check = 2*x + y + z
eq2_check = x + 3*y + 2*z
eq3_check = x + 2*y + 4*z

print(f"Equation 1 check (2x + y + z): {eq1_check:.4f} (Expected: 10)")
print(f"Equation 2 check (x + 3y + 2z): {eq2_check:.4f} (Expected: 13)")
print(f"Equation 3 check (x + 2y + 4z): {eq3_check:.4f} (Expected: 18)")

# Using np.allclose for more robust floating point comparison
print(f"All equations satisfied: {np.allclose([eq1_check, eq2_check, eq3_check], [10, 13, 18])}")
```

### Determinant of A and its Relationship to Unique Solution

```python
# Calculate the determinant of A_system
determinant_A_system = np.linalg.det(A_system)
print(f"Determinant of A_system: {determinant_A_system:.4f}")

# Explain the relationship
if np.isclose(determinant_A_system, 0):
    print("\nSince the determinant is approximately zero, the matrix A is singular. This means the system of equations either has no solution or infinitely many solutions. It does not have a unique solution.")
else:
    print("\nSince the determinant is non-zero, the matrix A is non-singular (invertible). This guarantees that the system of linear equations has a unique solution.")
```

## Question 5: Linear Algebra in a Simple AI Model

```python
# Define the feature matrix X, weight vector w, and bias b
X = np.array([[2, 3], [4, 5], [6, 7], [8, 9]])
w = np.array([0.5, 1.2])
b = 2

print("Feature Matrix X:\n", X)
print("\nWeight Vector w:", w)
print("Bias b:", b)
```

### Calculate Model Output (y = Xw + b)

```python
# Calculate the model output using y = Xw + b
model_output = np.dot(X, w) + b

print("Model Output (predicted values y):\n", model_output)
```

### Display Predicted Values

```python
# Display the predicted value for each sample
for i, prediction in enumerate(model_output):
    print(f"Sample {i+1} Predicted Value: {prediction:.4f}")
```

### Change Weight Values and Observe Prediction Changes

```python
# Change the weight values
new_w = np.array([1.0, 0.8])
new_b = 0.5 # Also change bias for more noticeable change

print("\nNew Weight Vector w:", new_w)
print("New Bias b:", new_b)

# Recalculate model output with new weights and bias
new_model_output = np.dot(X, new_w) + new_b

print("\nNew Model Output (predicted values y):\n", new_model_output)

# Display new predicted values
for i, prediction in enumerate(new_model_output):
    print(f"Sample {i+1} New Predicted Value: {prediction:.4f}")

print("\nAs you can see, changing the weights and bias significantly alters the predicted values.")
```
