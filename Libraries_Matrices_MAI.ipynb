**Question 2: Numerical Computing using NumPy**
- Create a 5 × 5 matrix containing random integers.
- Compute the Sum, Mean, Median, Standard Deviation, Maximum, and Minimum values.
- Reshape the matrix into a one-dimensional array.
- Find all elements greater than the average value.

```python
array = np.round(np.random.random((5,5))*100)
array
```

```python
print(np.sum(array))
print(np.mean(array))
print(np.median(array))
print(np.std(array))
print(np.max(array))
print(np.min(array))
```

```python
array.reshape(-1)
```

```python
array.ravel()
```

```python
GTA = array[array > np.mean(array)]
print(GTA)
```

**Question 4: Data Visualization using Matplotlib**
- Using the same dataset, generate a Line Plot, Bar Chart, Histogram, Scatter Plot, and Box Plot.

```python
# Line Plot: Sepal Length
plt.figure(figsize=(10, 6))
plt.plot(Iris.index, Iris['SepalLengthCm'])
plt.title('Line Plot of Sepal Length')
plt.xlabel('Index')
plt.ylabel('SepalLengthCm')
plt.grid(True)
# plt.show()
```

```python
# Bar Chart: Count of each species
plt.figure(figsize=(8, 5))
species_counts = Iris['Species'].value_counts()
plt.bar(species_counts.index, species_counts.values, color=['cyan', 'grey', 'red'])
plt.title('Count of Each Iris Species')
plt.xlabel('Species')
plt.ylabel('Count')
plt.show()
```

```python
# Histogram: Distribution of Sepal Length
plt.figure(figsize=(8, 5))
plt.hist(Iris['SepalLengthCm'], bins=10, edgecolor='black', color="grey")
plt.title('Sepal Length distribution')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.show()
```

```python
# Scatter Plot using Seaborn
plt.figure(figsize=(10, 6))
sns.scatterplot(x='SepalLengthCm', y='PetalLengthCm', hue='Species', data=Iris, s=100)
plt.title('Scatter Plot of Sepal Length vs Petal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.grid(True)
plt.show()
```

```python
# Box Plot: Sepal Width distribution by Species
# plt.figure(figsize=(10, 6))
sns.boxplot(x='Species', y='SepalWidthCm', data=Iris, palette='pastel')
plt.title('Box Plot of Sepal Width by Species')
plt.xlabel('Species')
plt.ylabel('Sepal Width (cm)')
plt.grid(axis='y')
plt.show()
```
